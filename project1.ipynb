```python
import pandas as pd
```

```python
data = {"product":["Laptop","Mouse","Keyboard","Monitor","Laptop","Mouse","Keyboard","Monitor"],
        "Region":["Athens","Patras","Athens","Thessaloniki","Patras","Athens","Patras","Athens"],
        "Sales":[1200,50,80,400,1100,60,90,450],
        "Profit":[300,15,25,100,280,20,30,120],
        "Quantity":[2,5,3,2,2,6,4,3]}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

Question:Which product and region has the highest sales?

```python
df.sort_values("Sales",ascending=False)
```

```python
df.groupby(["product","Region"])["Sales"].sum().plot(kind="bar")
```

Conclusion:The Laptop in Athens has the highest sales, with $1,200

Question:Which region has the highest total sales?

```python
df.groupby("Region")["Sales"].sum()
```

```python
df.groupby("Region")["Sales"].sum().plot(kind="bar")
```

Conclusion:Athens has the highest total sales, with $1,790

Question: Which product has the highest total sales?


```python
df.groupby("product")["Sales"].sum()
```

```python
df.groupby("product")["Sales"].sum().plot(kind="bar")
```

Conclusion:Keyboard has the highest total sales, with $2,300

Question:Which product has the highest total profit?

```python
df.groupby("product")["Profit"].sum()
```

```python
df.groupby("product")["Profit"].sum().plot(kind="bar")
```

Conclusion:The Laptop has the highest total profit, with $580

OVERALL CONCLUSION

Based on the analysis, we can see that Laptop has the highest total sales and the highest total profit. Athens is the region with the highest total sales. Overall, the analysis shows that sales and profit vary across differnt products and regions, providing useful insights into sales performance
